```python
import os
import numpy as np
from PIL import Image
```

Procedure untuk Ekstraksi Data Training

```python
def EkstraksiDataTraining(folder_training):
    # log ekstrasi data
    log = open("log.txt", "a")
    log.write("MENGEKSTRAKSI DATA TRAINING..\n")

    # buat isi ArrayDataLatih kosong
    ArrayDataLatih = []
    # buat isi LabelDataLatih kosong
    LabelDataLatih = []
    # buka folder training
    for label in os.listdir(folder_training):
        # tentukan label gambar (X atau Y)
        folder = os.path.join(folder_training, label)
        # buka folder X atau Y
        if os.path.isdir(folder):
            # untuk setiap gambar di folder X atau Y
            for filename in os.listdir(folder):
                # cek apakah file adalah gambar
                if filename.endswith(".jpg"):
                    
                    # log baca gambar
                    log.write("Membaca gambar : " + filename + "\n")
                    
                    # baca gambar
                    img = Image.open(os.path.join(folder, filename))
                    # ubah menjadi grayscale
                    img = img.convert("L")
                    # baca seluruh pixel
                    img_array = np.array(img)
                    # ubah menjadi biner, untuk setiap biner di baris
                    for i in range(len(img_array)):
                        # untuk setiap biner di kolom 
                        for j in range(len(img_array[i])):
                            # jika pixel < 128 maka 0, jika >= 128 maka 1
                            if img_array[i][j] < 128:
                                img_array[i][j] = 0
                            else:
                                img_array[i][j] = 1
                       
                    #  pemberian ketebalan
                    tebal = img_array.copy()
                    for i in range(1, len(img_array)-1):
                        for j in range(1, len(img_array[0])-1):
                            if img_array[i][j] == 0:
                                tebal[i+1][j] = 0                              
                    img_array = tebal
                
                    # susun menjadi satu baris
                    biner = img_array.flatten().astype(int)
                    # simpan data
                    ArrayDataLatih.append(biner)
                    # simpan label
                    LabelDataLatih.append(label)
                    
    # log total data   
    log.write("Jumlah Data Training : " + str(len(ArrayDataLatih)) + "\n\n")
    log.close()
    
    return ArrayDataLatih, LabelDataLatih
```

Procedure untuk Ekstraksi Data Testing

```python
def EkstraksiDataTesting(folder_testing):
    # log ekstrasi data
    log = open("log.txt", "a")
    log.write("MENGEKSTRAKSI DATA TESTING.. \n")

    # buat isi ArrayDataTesting kosong
    ArrayDataTesting = []
    # buat isi LabelDataTesting kosong
    LabelDataTesting = []
    # NamaFileTesting = []
    # buka folder testing
    for label in os.listdir(folder_testing):
        # tentukan label gambar (X atau Y)
        folder = os.path.join(folder_testing, label)
        # buka folder X atau Y
        if os.path.isdir(folder):
            # untuk setiap gambar
            for filename in os.listdir(folder):
                # cek apakah file adalah gambar
                if filename.endswith(".jpg"):
                    
                    # log baca gambar
                    log.write("Membaca gambar : " + filename + "\n")
                    
                    # baca gambar
                    img = Image.open(os.path.join(folder, filename))
                    # ubah menjadi grayscale
                    img = img.convert("L")
                    # baca seluruh pixel
                    img_array = np.array(img)
                    # ubah menjadi biner, untuk setiap biner di baris
                    for i in range(len(img_array)):
                        #  untuk setiap biner di kolom
                        for j in range(len(img_array[i])):
                            # jika pixel < 128 maka 0, jika >= 128 maka 1
                            if img_array[i][j] < 128:
                                img_array[i][j] = 0
                            else:
                                img_array[i][j] = 1
                           
                    # pemberian ketebalan  
                    tebal = img_array.copy()
                    for i in range(1, len(img_array)-1):
                        for j in range(1, len(img_array[0])-1):
                            if img_array[i][j] == 0:
                               tebal[i+1][j] = 0
                    img_array = tebal
                    
                    # susun menjadi satu baris
                    biner = img_array.flatten().astype(int)
                    # simpan data
                    ArrayDataTesting.append(biner)
                    # simpan label
                    LabelDataTesting.append(label)
                    # NamaFileTesting.append(filename)

    # log total data   
    log.write("Jumlah Data Testing : " + str(len(ArrayDataTesting)) + "\n\n")
    log.close()
        
    return ArrayDataTesting, LabelDataTesting
```

Function untuk menghitung jarak antara data testing dan tiap Data Latih

```python
def HitungJarakAntDataTestingDataLatih(data_testing, ArrayDataLatih, LabelDataLatih):
     # log  hitung jarak 
    log = open("log.txt", "a")
    log.write("=======================================\n")
    log.write("MENGHITUNG JARAK..\n")
    
    hasil_jarak = []
    # ambil 1 data ArrayDataTesting
    # untuk setiap data di ArrayDataLatih
    for i in range(len(ArrayDataLatih)):
        # ambil 1 data ArrayDataLatih
        data_latih = ArrayDataLatih[i]
        jarak = 0
        # unutuk setiap 1 data di ArrayDataTesting
        for j in range(len(data_testing)):
            # Hitung hitung mutlak dari selisih antara data_testing dan data_latih
            selisih = abs(int(data_testing[j]) - int(data_latih[j]))
            jarak = jarak + selisih
        # simpan hasil jarak
        hasil_jarak.append([jarak, LabelDataLatih[i]])
        
        # log hitung jarak 
        log.write(str(i+1) + ". Label : " + LabelDataLatih[i] + " | Jarak : " + str(jarak) + "\n")
    log.close()

    return hasil_jarak
```

Procedure untuk Sorting Hasil Jarak 

```python
def SortingHasilJarak(hasil_jarak):
    hasil_sorting = sorted(hasil_jarak)
    
    # log hasil sorting
    log = open("log.txt","a")
    log.write("\nHASIL SORTING\n")
    for i in range(len(hasil_sorting)):
        log.write(str(i+1) + ". Label : " + hasil_sorting[i][1] + " | Jarak : " + str(hasil_sorting[i][0]) + "\n")
    log.close()
    
    return hasil_sorting
    
```

Function untuk Hitung Akurasi

```python
def HitungAkurasi(LabelDataTesting, LabelPrediksi):
   jumlah_prediksi_benar=0
   # untuk semua data di data latih
   for i in range(len(LabelDataTesting)):
      # jika prediksi sama dengan label (benar)
      if LabelPrediksi[i] == LabelDataTesting[i]:
         # maka prediksi benar
         jumlah_prediksi_benar += 1
   # hitung akurasi
   akurasi = (jumlah_prediksi_benar / (len(LabelDataTesting))) * 100
   
   # log hasil akurasi
   log = open("log.txt","a")
   log.write("\nHASIL AKURASI\n")
   log.write("Prediksi Benar : " + str(jumlah_prediksi_benar) + "\n")
   log.write("Akurasi : " + str(akurasi) + "%\n")
   log.close()
    
   return akurasi
```

Main Program

```python
ArrayDataLatih, LabelDataLatih, NamaFileTraining = EkstraksiDataTraining("dataset/training")
ArrayDataTesting, LabelDataTesting, NamaFileTesting = EkstraksiDataTesting("dataset/testing")
LabelPrediksi = []
    
for i in range(len(ArrayDataTesting)):
    hasil_jarak = HitungJarakAntDataTestingDataLatih(ArrayDataTesting[i], ArrayDataLatih, LabelDataLatih)
    # mengurutkan data testing verdasarkan hasil_jarak terkecil ke terbesar
    hasil_sorting = SortingHasilJarak(hasil_jarak)
    # ambil tetangga terdekat, k = 1 (data pertama terdekat)
    prediksi = hasil_sorting[0][1]
    # simpan prediksi
    LabelPrediksi.append(prediksi)
    print("=======================================") 
    print("Data Testing :", i+1 ) 
    # print("Nama File :", NamaFileTesting[i]) 
    print("Label Asli :", LabelDataTesting[i]) 
    print("\nHasil Sorting") 
    for j in range(len(hasil_sorting)): 
        print(j+1, hasil_sorting[j]) 
    print("\nPrediksi :", prediksi) 
    if prediksi == LabelDataTesting[i]: 
        print("Status : Benar") 
    else: 
        print("Status : Salah")
        
     # log prediksi
    log = open("log.txt", "a")
    log.write("-------------------------\n")
    log.write("Data Testing : " + str(i+1) + "\n")
    # log.write("Nama File    : " + NamaFileTesting[i] + "\n")
    log.write("Label Asli   : " + LabelDataTesting[i] + "\n")
    log.write("Prediksi     : " + prediksi + "\n")
    if prediksi == LabelDataTesting[i]:
        log.write("Status       : BENAR\n")
    else:
        log.write("Status       : SALAH\n")
    log.write("=======================================\n")
    log.write("\n")
    log.close()
        
akurasi = HitungAkurasi(LabelDataTesting, LabelPrediksi)
print("\nAkurasi :", akurasi, "%")
```
